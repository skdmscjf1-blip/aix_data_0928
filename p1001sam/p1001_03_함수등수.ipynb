```python
import pandas as pd
df = pd.read_excel('file/score.xlsx', index_col='지원번호')
```

```python
# 컬럼전체 이름변경
df['학교'] = df['학교']+'등학교'
# df.replace({'고':'고등학교'}, inplace=True)
# df['학교'] = df['학교'].str.replace({'고':'고등학교'})
```

```python
# 키 + cm 추가
# df['키'] = df['키'] + 'cm' #df['키']:int, 'cm':str
# df['키'] = df['키'] + 100  # int+int

# 타입이 다른 경우, 타입변경후 적용
df['키'] = df['키'].astype(str) + 'cm'  # str으로 변환
```

```python
# 함수 적용 : apply() - 컬럼 전체에 적용
def add_cm(h):
    return str(h)+'센티미터'

df['키'] = df['키'].apply(add_cm)
```

```python
# 함수적용
def capChange(lang):
    if pd.notnull(lang):
        return lang.lower()
    else:
        return "특기없음"
    
df['SW특기'] = df['SW특기'].apply(capChange)
df['SW특기']

# df['SW특기'].str.lower()
```

```python
df = pd.read_excel('file/score.xlsx', index_col='지원번호')
```

```python
# 1.합계 : 과목총점을 계산해서 컬럼을 추가하시오.
# 2.평균 컬럼을 추가하시오.
# 3.등급 컬럼을 추가해서 90점이상 A, 80점이상 B, 70점 C, 60점 D, F 를 입력하시오.
# 함수를 사용할것
df['합계'] = df['국어'] + df['영어'] + df['수학'] + df['과학'] + df['사회']
df['평균'] = df['합계'] / 5
df['등급'] = 'F'
```

```python
def cal_rank(s):
    if s>=90: return 'A'
    elif s>=80: return 'B'
    elif s>=70: return 'C'
    elif s>=60: return 'D'
    else: return 'F'

df['등급'] = df['평균'].apply(cal_rank)
df
```

```python
# 등수 : ascending=False 역순정렬, method='min' 1,2,2,4
df['등수'] = df['합계'].rank(ascending=False).astype(int)
df
```
