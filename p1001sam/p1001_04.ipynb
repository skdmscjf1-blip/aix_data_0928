```python
import pandas as pd
df = pd.read_excel('file/score.xlsx',index_col='지원번호')
```

```python
# 합계 컬럼 추가
df['합계'] = df['국어'] + df['영어'] + df['수학']
# SW특기 있으면 합계점수 5점을 추가해서 입력하시오.
# ( 함수를 사용해서 적용할 것 )
```

```python
def cal_sw(df2):
    # if df2['SW특기'] != '':
    if pd.notnull(df2['SW특기']):
        return df2['합계']+5
    else:
        return df2['합계']
    
df['합계'] = df.apply(cal_sw,axis=1)
```
