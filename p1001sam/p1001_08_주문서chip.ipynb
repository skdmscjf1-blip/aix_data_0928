```python
# 1. item_name 당 주문서 개수를 출력하시오. 10개만 출력
# 2. item_name 당 주문 총량(quantity)을 출력하시오.
# 3. 가장 많이 주문한 item_name  10개를 출력하시오.

# 4. 주문당 평균 금액을 출력하시오.
# 5. 한 주문에 10달러 이상 사용한 주문 번호(id) 출력하시오.
# Chicken Bowl을 2개 이상 구매한 주문서 출력하시오.
```

```python
import pandas as pd
df = pd.read_csv('file/chipotle.tsv', sep = '\t')
```

```python
# ============================================================
# 1. item_name 당 주문서 개수 출력 - 10개만
# groupby : 같은 값끼리 묶기
# count : 개수 세기
# head(10) : 앞에서 10개
# ============================================================

# df.groupby('item_name')['order_id'].count().head(10)


# ============================================================
# 2. item_name 당 주문 총량(quantity)
# quantity : 수량
# sum : 합계
# ============================================================

# df.groupby('item_name')['quantity'].sum().sort_values(ascending=False)


# ============================================================
# 3. 가장 많이 주문한 item_name 10개
# sort_values : 값 정렬
# ascending=False : 큰 값부터 정렬
# ============================================================

# df.groupby('item_name')['quantity'].sum().sort_values(ascending=False).head(10)


# ============================================================
# 4. 주문당 평균 금액
# replace : $ 제거
# astype(float) : 숫자(실수)로 변경
# mean : 평균
# ============================================================

df['item_price'] = df['item_price'].str.replace('$', '')
df['item_price'] = df['item_price'].astype(float)
# df.groupby('order_id')['item_price'].sum()
order_price = df.groupby('order_id')['item_price'].sum()

# order_price.mean()


# ============================================================
# 5. 한 주문에 10달러 이상 사용한 주문번호(id)
# >= 10 : 10 이상인 데이터 찾기
# index : 주문번호(order_id) 가져오기
# ============================================================

order_price[order_price >= 10].index


# ============================================================
# 6. Chicken Bowl을 2개 이상 구매한 주문서
# & : 두 조건을 모두 만족
# ============================================================

# df[(df['item_name'] == 'Chicken Bowl') & (df['quantity'] >= 2)]
```
