### 문자열 함수 .str

```python
# 문자열함수 str함수 - startswith:강으로 시작하는 문자열
# startswith,endswith,contains,lower,upper,capitalize
# replace - 대체
# split-분리,strip-공백제거,find-찾기,-1,len:글자수

# df[ df['이름'].str.startswith('강') ]
# df.loc[df['이름'].str.startswith('강')]

# contains 포함되어 있는 글자, ~ : not
# df[~df['이름'].str.contains('열')]

# lower,upper,capitalize
df['SW특기']
df['SW특기'].str.lower()
df['SW특기'].str.upper()
df['SW특기'].str.capitalize()

# isin : 여러개 검색 ['python','java']
langs = ['python','java']
# df[df['SW특기'].str.lower().isin(langs)] #SW특기 값 소문자로 변경후 리스트검색

# 
# df['SW특기'].str.lower().contains()
df['SW특기']
df['SW특기'].str.contains('Java',na=False)
df[df['SW특기'].str.contains('Java',na=True)]  #na=True : NaN 데이터 출력
df[df['SW특기'].str.contains('Java',na=False)]  #na=False : NaN 출력하지 않음
df[df['SW특기'].str.lower().str.contains('java',na=True)] # 소문자로 변경후 검색

# unique 중복제거
df['학교']
df['학교'].unique()
```
