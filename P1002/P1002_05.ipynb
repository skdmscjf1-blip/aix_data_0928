```python
df = pd.read_excel('file/score.xlsx',index_col='지원번호')
```

```python
# 선 스타일
x = [1,2,3]
y = [20,40,80]
y2 = [50,10,30]
y3 = [10,20,50]

plt.title('꺾은선 graph')  # 상단타이틀
# figure : 그래프 사이즈 변경 , figsize : 크기 , dpi : 해상도
# facecolor : 배경색변경
plt.plot(x,y,label='국어',linewidth=2,marker='o',ms=10,mfc='yellow',linestyle='-',\
        color='pink')
plt.plot(x,y2,label='영어',linewidth=2,marker='x',ms=10,mfc='yellow',linestyle=':',\
        color='blue')
plt.plot(x,y3,label='수학',linewidth=2,marker='^',ms=10,mfc='yellow',linestyle='-.',\
        color='green')
# 그래프 안에 글자를 입력 : plt.text
# plt.text(x[0],y[0]+0.3,2,ha='center',color='red')
plt.legend()
for i,v in enumerate(y) :
    plt.text(x[i],y[i]+2,v,ha='center',color='red')
for i,v in enumerate(y2) :
    plt.text(x[i],y2[i]+2,v,ha='center',color='red')
for i,v in enumerate(y3) :
    plt.text(x[i],y3[i]+2,v,ha='center',color='red')
```

```python
x = df.index
y = df['국어'] #타입이 pandas타입 -> y[0],y[1]... 불가 
# y = list(df['국어']) # 리스트타입 -> y[0],y[1]... 가능
plt.figure(figsize=(10,5))
plt.plot(x,y+2,label='국어점수')
# plt.bar(x,y,label='국어점수')
for i,v in enumerate(y) :
    plt.text(x[i],v+2,v,ha='center')
    # plt.text(x[i],y[i],v,ha='center')
plt.legend()
plt.show()
```
