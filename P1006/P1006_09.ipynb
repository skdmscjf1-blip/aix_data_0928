```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['axes.unicode_minus'] = False     # - 부호처리
matplotlib.rcParams['font.family'] = 'Malgun Gothic'  # 한글처리 : 윈도우 맑은고딕 저장
matplotlib.rcParams['font.size'] = 15                 # 전체 글자폰트크기 조정
# matplotlib.rcParams['font.family'] = 'AppleGothic'    # 맥컴 애플고딕 저장
import pandas as pd
import numpy as np
import re                   # 정규표현식
```

```python
df = pd.read_csv('file/namu2.csv')
df = df.head(5)
```

```python
# 글자에 대한 전처리 작업 - 한글이외 글자 삭제 , 필요없는 글자 삭제
def text_cleaning(text):
    if pd.isna(text):
        return ''
    hangul = re.compile('[^ ㄱ-ㅣ가-힣]+')   #정규표현식에 해당이 되는지 확인.
    return hangul.sub('',text)
```

```python
df['content'] = df['content'].apply(lambda x : text_cleaning(x))
df['content']
df_content_all = ' '.join(df['content']) #리스트에 있는 것을 1개의 문자열로 합침.
df_content_all
```

```python
from konlpy.tag import Okt #한글형태소 분석
from collections import Counter
import os
os.environ["JAVA_HOME"] = r"C:\jdk21"
```

```python
n_tag = Okt()
nouns = n_tag.nouns(df['content'][0]) #명사로 추출
count = Counter(nouns)             #명사로 추출한 단어 개수 확인
count   #딕셔너리 타입
```

```python
#1개 단어 키워드 제외 - 딕셔너리 타입
# count #딕셔너리 타입
remove_counter = Counter({x:count[x] for x in count if len(x)>1})
remove_counter
```

```python
remove_counter['편집']
```

```python
#욕설
#불용어 
stopwords = ['상위','편집', '문서', '내용', '누설', '아래', '해당', '설명', '표기', '추가', '모든', '사용', '매우', '가장','줄거리', '요소', '상황', '편집', '틀', '경우', '때문', '모습', '정도', '이후', '사실', '생각', '인물','이름', '년월']
remove_counter2 = Counter({x : remove_counter[x] for x in remove_counter if x not in stopwords})
remove_counter2
```
