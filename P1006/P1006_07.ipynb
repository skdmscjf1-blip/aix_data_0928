```python
data = []
for url in df['url'].head(50):
    # # 2. selenium : 자동화 도구 - 스크롤없이 가져옴
    # namu_url = url
    # options = Options()
    # options.add_experimental_option("excludeSwitches", ["enable-automation"])
    # options.add_experimental_option("useAutomationExtension", False)
    # options.add_argument("--disable-blink-features=AutomationControlled")
    # browser = webdriver.Chrome(options=options)
    # browser.maximize_window() # 화면 최대창 확대
    # browser.get(namu_url)
    # time.sleep(2)
    res = requests.get(url)
    time.sleep(1)
    soup = BeautifulSoup(res.text,'lxml')
    title_tag = soup.find('a',{'class':'J23-5aq4'})
    title = title_tag.get_text(strip=True) if title_tag else None
    category_tag = soup.find('ul',{'class':'EVpyuXop'})
    category = category_tag.get_text(strip=True) if category_tag else None
    content_tag = soup.find('div',{'class':'_6YZgjHNT Oud4u5fe'})
    content = content_tag.get_text(strip=True) if content_tag else None
    data.append({
        'title':title,
        'category':category,
        'content':content
    })
print("완료됨.")
# # # 파일저장
# soup = BeautifulSoup(browser.page_source,'lxml')
# with open('file/namu_con1.html','w',encoding='utf-8') as f:
#     f.write(soup.prettify())
```

```python
n_df = pd.DataFrame(data)
n_df.to_csv('file/namu2.csv')
```
