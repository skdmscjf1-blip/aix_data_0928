```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['axes.unicode_minus'] = False     # - 부호처리
matplotlib.rcParams['font.family'] = 'Malgun Gothic'  # 한글처리 : 윈도우 맑은고딕 저장
matplotlib.rcParams['font.size'] = 15                 # 전체 글자폰트크기 조정
# matplotlib.rcParams['font.family'] = 'AppleGothic'    # 맥컴 애플고딕 저장
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel('file/students.xlsx')
# df.groupby('학년')['학교'].value_counts()
school= df['학교'].value_counts(normalize=True)*100 # 퍼센트로 나옴
explode1 = [0.5,0.1,0] #원형 사이간격 띄우기 
colors = ['#ffadad']
school.index
school.values
plt.pie(school.values,labels=school.index,startangle=90,counterclock=False,explode=explode1,autopct='%.1f%%')
# plt.pie(values,labels=labels,startangle=90,counterclock=False,explode=explode,autopct='%.1f%%')
plt.show()
```
