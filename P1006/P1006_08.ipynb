```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['axes.unicode_minus'] = False     # - 부호처리
matplotlib.rcParams['font.family'] = 'Malgun Gothic'  # 한글처리 : 윈도우 맑은고딕 저장
matplotlib.rcParams['font.size'] = 15                 # 전체 글자폰트크기 조정
# matplotlib.rcParams['font.family'] = 'AppleGothic'    # 맥컴 애플고딕 저장
import pandas as pd
import numpy as np
import re                   # 정규표현식
```

```python
df = pd.read_csv('file/namu2.csv')
```

```python
# 글자에 대한 전처리 작업 - 한글이외 글자 삭제 , 필요없는 글자 삭제
def text_cleaning(text):
    if pd.isna(text):
        return ''
    hangul = re.compile('[^ ㄱ-ㅣ가-힣]+')   #정규표현식에 해당이 되는지 확인.
    return hangul.sub('',text)
```

```python
df['title'] = df['title'].astype(str)
```

```python
#한글 이외 모든 문자 삭제
df['title'] = df['title'].apply(lambda x : text_cleaning(x))
df['category'] = df['category'].apply(lambda x : text_cleaning(x))
df['content'] = df['content'].apply(lambda x : text_cleaning(x))
```

```python
#각 컬럼마다 컬럼을 모두 1개의 문자열로 변환
titles = ' '.join(df['title'])
categorys = ' '.join(df['category'])
contents = ' '.join(df['content'])
contents
```
