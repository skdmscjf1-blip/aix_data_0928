```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['axes.unicode_minus'] = False     # - 부호처리
matplotlib.rcParams['font.family'] = 'Malgun Gothic'  # 한글처리 : 윈도우 맑은고딕 저장
matplotlib.rcParams['font.size'] = 15                 # 전체 글자폰트크기 조정
# matplotlib.rcParams['font.family'] = 'AppleGothic'    # 맥컴 애플고딕 저장
import pandas as pd
import numpy as np
```

```python
df=pd.read_excel('file/score.xlsx',index_col='지원번호')
```

```python
# alist = [1+1,2+1,3+1]
# alist+1
arr = np.arange(8)  # numpy명령어
# arr
arr*3
```

```python
x = df['이름']
y = df['국어']
y2 = df['영어']
y3 = df['수학']
# 강태원-0.2 :안됨.
plt.bar(arr-0.3,y,width=0.3,label='국어')
plt.bar(arr,y2,width=0.3,label='영어')
plt.bar(arr+0.3,y3,width=0.3,label='수학')
plt.legend(ncol=3,loc='upper center') # 범례 위치
# plt.plot(x,y,marker='o')
# plt.plot(x,y2)
# plt.yticks([0,10,20,90,100,110]) #눈금자 수동표시
plt.ylim(0,130)  # 눈금자 처음,끝표시
# plt.xticks(x,df.index) # x축 눈금자 수동표시
# plt.xticks(arr,df.index) # x축 눈금자 수동표시
plt.xticks(arr,x) # x축 눈금자 수동표시
# plt.xticks(x,[1,2,3,4,5,6,7,8])
for i,(v,v2,v3) in enumerate(zip(df['국어'],df['영어'],df['수학'])):
    plt.text(i-0.3,v+2,v,ha='center')  #i:x좌표, v:y좌표, v:값, ha:가운데정렬
    plt.text(i,v2,v2,ha='center')  #i:x좌표, v:y좌표, v:값, ha:가운데정렬
    plt.text(i+0.3,v3,v3,ha='center')  #i:x좌표, v:y좌표, v:값, ha:가운데정렬
plt.show()
```
