```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['axes.unicode_minus'] = False     # - 부호처리
matplotlib.rcParams['font.family'] = 'Malgun Gothic'  # 한글처리 : 윈도우 맑은고딕 저장
matplotlib.rcParams['font.size'] = 15                 # 전체 글자폰트크기 조정
# matplotlib.rcParams['font.family'] = 'AppleGothic'    # 맥컴 애플고딕 저장
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel('file/연령별인구현황01.xlsx',skiprows=3,nrows=1,usecols='B,E:Y',index_col='행정기관')
# 남자데이터
m_df = pd.read_excel('file/연령별인구현황01.xlsx',skiprows=3,nrows=1,usecols='B,AB:AV',index_col='행정기관')
# 여자데이터
w_df = pd.read_excel('file/연령별인구현황01.xlsx',skiprows=3,nrows=1,usecols='B,AY:BS',index_col='행정기관')

```

```python
m_df.columns = df.columns
w_df.columns = df.columns
```

```python
# 막대그래프 출력하시오.
# 왼쪽 여자막대그래프/ 오른쪽 남자막대그래프
# df['0~4세'] = df['0~4세'].str.replace(',','').astype(int) #1개만 변경
```

```python
m_df2 = m_df.iloc[0].str.replace(',','').astype(int)
w_df2 = w_df.iloc[0].str.replace(',','').astype(int)
x = m_df2.index
y = m_df2.values
x2 = w_df2.index
y2 = w_df2.values

plt.barh(x,y)
plt.barh(x2,-y2)
plt.title('2025년 대한민국 인구 피라미드')
plt.show()

```
