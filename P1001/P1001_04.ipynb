```python
import pandas as pd
df = pd.read_excel('file/score.xlsx',index_col='지원번호')
```

```python
# 합계 컬럼 추가
# SW특기 있으면 합계점수 5점 추가해서 입력하시오.
# 함수를 사용해서 적용할것

df['합계'] = df['국어']+df['영어']+df['수학']+df['과학']+df['사회']
# type(df['SW특기'].values[4])
df
```

```python
def cal_sw(df2) :
    # if df2['SW특기'] != '':
    if pd.notnull(df2['특기']) :
        return df2['합계'] +5
    else :
        return df2['합계']
df['합계'] = df.apply(cal_sw,axis=1)
```

```python
def cal_sw(lang) : 
    if pd.notnull(lang) :
        return lang.lower()
    else : 
        return '없음'

df['SW특기'] = df['SW특기'].apply(cal_sw)
```
