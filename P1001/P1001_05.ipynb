```python
# 0일 1월 2화 3수 4목 5금 6토

def day(today) :
    if today == 0: return '일요일'
    elif today == 1: return '월요일'
    elif today == 2: return '화요일'
    elif today == 3: return'수요일'
    elif today == 4: return '목요일'
    elif today == 5: return'금요일'
    elif today == 6: return '토요일'

df['요일'] = df['요일'].apply(day)
```

```python
data = {
    '번호':['1번','2번','3번','4번','5번','6번','7번','8번'],
    '이름':['홍길동','유관순','이순신','강감찬','김구','김유신','홍길자','홍길순'],
    '달러':['$19,700','$18,400','$16,800','$18,700','$18,800','$20,200','$18,800','$19,000'],
    '우대':[90,40,80,40,15,80,55,100]
}

df = pd.DataFrame(data)
df.set_index('번호',inplace=True)
```

```python
df
```

```python
df['금액'] = 0


#우대 50이상인 사람에게 달려+달러X우대X0.01
#우대 50미만 달러

#달러 -> int 변경($,쉼표 제거)
df['달러'] = df['달러'].str.replace('$','')
df['달러'] = df['달러'].str.replace(',','').astype(int)
df['달러']

def dal(df2):
    if df2['우대'] >= 50:
        df2['금액'] = df2['달러']+(df2['달러']*df2['우대']*0.01)
        return df2['금액']
    else :
        return df2['달러']

df['금액'] = df.apply(dal,axis=1)
```
