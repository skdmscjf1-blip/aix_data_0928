```python
import pandas as pd
df =pd.read_excel('file/score.xlsx',index_col='지원번호')
```

```python
#합계,평균,결과 컬럼을 추가
#결과는 평균점수가 70점 이상은 합격 , 불합격으로 출력하시오.
df['합계'] = 0
df['평균'] = 0
df['결과'] = '불합격'
df['합계'] = df['국어']+df['영어']+df['수학']
df['평균'] = df['합계']/3
df.loc[df['평균']>=70,'결과'] = '합격' 
```

```python
#컬럼순서변경

# df2 = df[['이름','평균','결과']]
# df2
df = df[['이름','평균','결과','학교', '학년', '키', '국어', '영어', '수학', 'SW특기', '합계']]
```

```python
#[cols 리스트 순서를 변경]
#1.리스트 타입으로 변경후 순서를 변경
cols = list(df.columns)

cols[-2:] #평균 , 결과 컬럼
type(cols[0]) # str -> list 타입으로 변경하려면 [] 로 묶어야함
cols[1:-2] # 타입 list 이름~ 합계 
# cols[0]+cols[-2:]+cols[1:-2] # str + list + list => error 발생
[cols[0]]+cols[-2:]+cols[1:-2] # list + list + list
cols = [cols[0]]+cols[-2:]+cols[1:-2]
#3. 변경된 리스트 적용
df=df[cols]
df
```

```python
# 국어,영어,수학,과학,사회 합계 컬럼을 추가하고
# 합계점수가 350점이상만 출력하시오
# 350점 미만 삭제후 출력

df['합계'] = df['국어']+df['영어']+df['수학']+df['과학']+df['사회']
total = df['합계']>350
df2 = df[~total].index
df.drop(index=df2,inplace=True)
```
