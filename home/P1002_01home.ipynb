```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['axes.unicode_minus'] = False # '-' 마이너스 부호 처리
matplotlib.rcParams['font.family'] = 'Malgun Gothic' # 한글처리 : 윈도우 맑은고딕 저장
matplotlib.rcParams['font.size'] = 15               # 전체 글자폰트크기 조정
# matplotlib.rcParams['font.family'] = 'AppleGothic' #맥컴 애플고딕 저장
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel('file/students.xlsx')
df
```

```python
school = df.groupby('학교')[['국어','영어','수학']].mean()
```

```python
school.index
school.columns
school.values
school
```

```python
x = school.index

y = school['국어']
y2 = school['영어']
y3 = school['수학']

a = np.arange(len(x))
plt.title('학생성적프로그램')
plt.bar(a-0.2,y,width=0.2,label='국어')
plt.bar(a,y2,width=0.2,label='영어')
plt.bar(a+0.2,y3,width=0.2,label='수학')
plt.xticks(a,x)
plt.yticks([0,50,80,100])
plt.legend(ncols=3)
for i,(v,v2,v3) in enumerate(zip(y,y2,y3)) : 
    plt.text(i-0.2 ,v+2 ,f'{v:.2f}',ha='center')
    plt.text(i     ,v2+1  ,f'{v2:.2f}',ha='center')
    plt.text(i+0.2 ,v3+1 ,f'{v3:.2f}',ha='center')
plt.show()
```
